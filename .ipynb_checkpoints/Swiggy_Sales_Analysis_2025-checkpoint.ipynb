# Charts Design

### Montly Sales Trend

```python
df['Order Date']=pd.to_datetime(df['Order Date'])

df["YearMonth"]=df["Order Date"].dt.to_period("M").astype(str) #2025-06

montly_revenue=df.groupby("YearMonth")["Price (INR)"].sum().reset_index()

# montly_revenue

plt.figure()
plt.plot(montly_revenue["YearMonth"],montly_revenue["Price (INR)"])
plt.xticks(rotation=45)
plt.xlabel("Month")
plt.ylabel("Total Sales (INR)")
plt.title("Monthly Sales Trend")
plt.tight_layout() 
plt.show()
```

### Daily Sales Trend

```python
df["DayName"]=pd.to_datetime(df['Order Date']).dt.day_name()  #  Sunday

daily_revenue=(
    df.groupby("DayName")["Price (INR)"].sum()
    .reindex(["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"])
)

# daily_revenue

plt.figure(figsize=(10,5))

bars = plt.bar(
    daily_revenue.index,
    daily_revenue.values,
    color=["green","red","blue","orange","purple","cyan","brown"]
)

# Add labels on top of each bar
plt.bar_label(bars, fmt="₹%.0f", padding=3)

plt.title("Daily Revenue Trend (Mon-Sun)")
plt.xlabel("Day")
plt.ylabel("Revenue (INR)")
plt.xticks(rotation=30)

plt.show()
```

### Total Sales by Food Type (Veg vs Non-Veg) 

```python
non_veg_keywords=[
    "chicken","egg","fish","mutton",
    "prawn","biryani","kabab","kebab",
    "non-veg","non veg"
]

# filter the non-veg or veg data
df["Food Category"]=np.where(
    df["Dish Name"].str.lower().str.contains("|".join(non_veg_keywords),na=False),
    "Non-Veg",
    "Veg"
)

# calculate the category-vise(veg/non-veg) price 
food_revenue=(
    df.groupby("Food Category")["Price (INR)"].sum()
    .reset_index()
)

fig=px.pie(
    food_revenue,
    values="Price (INR)",
    names="Food Category",
    hole=0.5,
    title="Revenue Contribution: Veg vs Non-Veg",    
)

# show the labels
fig.update_traces(
    textinfo="percent+label",
    pull=[0.05,0]
)

fig.update_layout(
    height=500,
    margin=dict(t=60,b=40,l=40,r=40)
)
fig.show()
```

### Total Sales by State

```python
fig=px.bar(
    df.groupby("State",as_index=False)["Price (INR)"].sum()
    .sort_values("Price (INR)",ascending=False),
    x="Price (INR)",
    y="State",
    orientation="h",
    title="Revenue by State (INR)"
)
fig.update_layout(height=600,yaxis=dict(autorange="reversed"))
fig.show()
```

### Quarterly Performance Summary 

```python
df["Order_Date"]=pd.to_datetime(df["Order Date"])

df["Quarter"]=df["Order_Date"].dt.to_period("Q").astype(str)

quartely_summary=(
    df.groupby("Quarter",as_index=False).agg(
        Total_sales=("Price (INR)","sum"),
        Avg_Rating=("Rating","mean"),
        Total_Orders=("Order_Date","count")
    ).sort_values("Quarter")
)

# this 2 line is optional when we see the round of-value
quartely_summary["Total_sales"]=quartely_summary["Total_sales"].round(0)
quartely_summary["Avg_Rating"]=quartely_summary["Avg_Rating"].round(2)

quartely_summary
```

### Top 5 Cities by Sales 

```python
top_5_cities=(
    df.groupby("City")["Price (INR)"].sum()
    .nlargest(5)
    .sort_values()
    .reset_index()
)

fig=px.bar(
    top_5_cities,
    x="Price (INR)",
    y="City"
    ,orientation="h",
    title="Top 5 Cities by Sales (INR)",
    color_discrete_sequence=["red"]
)
fig.show()
```

### Weekly Trend Analysis 

```python
df["Weekly"]=pd.to_datetime(df["Order Date"]).dt.to_period("W").astype(str)

fig=px.bar(
    df.groupby("Weekly")["Price (INR)"].sum().sort_values().reset_index(),
    x="Price (INR)",
    y="Weekly"
    ,orientation="h",
    title="Weekly Trend Analysis",
)
fig.update_layout(height=600)
fig.show()
```

```python
weekly_sales = (
    df.groupby("Weekly")["Price (INR)"]
      .sum()
      .reset_index()
)

fig = px.line(
    weekly_sales,
    x="Weekly",
    y="Price (INR)",
    title="Weekly Sales Trend",
    markers=True
)

fig.show()
```
